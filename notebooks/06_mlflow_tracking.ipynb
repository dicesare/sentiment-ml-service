# 06 — MLflow experiment tracking
A common tracking contract makes classical, LSTM and transformer candidates comparable.

```python
import mlflow

def log_candidate(name, params, metrics, artifact_path=None):
    with mlflow.start_run(run_name=name):
        mlflow.log_params(params)
        mlflow.log_metrics(metrics)
        if artifact_path:
            mlflow.log_artifact(artifact_path)
```

## Artifact policy
Log dataset version, code revision, seed, preprocessing configuration, metrics and model checksum. Keep local MLflow stores, TensorBoard logs and trained weights out of Git.
