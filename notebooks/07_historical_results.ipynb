# 07 — Historical experiment results
This notebook turns the executed outputs found across the original `dev` and `flask` branches into a traceable comparison. Values are transcribed from preserved outputs; they are not presented as a fresh benchmark.

```python
import pandas as pd
import matplotlib.pyplot as plt

results = pd.DataFrame([
 {'model':'BoW + logistic regression','accuracy':0.794777,'f1':None,'roc_auc':None,'source':'test_mod_simple'},
 {'model':'TF-IDF + logistic regression','accuracy':0.779415,'f1':None,'roc_auc':None,'source':'test_mod_simple'},
 {'model':'small BERT','accuracy':0.7613,'f1':0.7651,'roc_auc':0.8200,'source':'TensorFlow_Ananlyse'},
 {'model':'custom BERT','accuracy':0.7781,'f1':0.7839,'roc_auc':0.8295,'source':'Tensorflow_Classifieur'},
 {'model':'tuned LSTM','accuracy':0.6544,'f1':0.6409,'roc_auc':0.6545,'source':'test_pipeline_machine_learnia'}
])
results
```

```python
ax = results.set_index('model')[['accuracy','f1','roc_auc']].plot.barh(figsize=(10,5), xlim=(0,1))
ax.set(title='Historical test metrics', xlabel='score')
ax.grid(axis='x', alpha=.25)
plt.tight_layout()
```

## Reading the comparison
The sparse baselines are competitive with the transformer experiments on accuracy. Custom BERT has the strongest documented balance of F1 and ROC-AUC. The LSTM search produced very high recall in one evaluation but weak precision and inconsistent built-in accuracy, signalling threshold/calibration problems rather than a result to hide.

```python
bert_detail = pd.DataFrame([
 {'run':'small BERT','accuracy':.7613,'precision':.7518,'recall':.7789,'f1':.7651,'roc_auc':.8200},
 {'run':'custom BERT','accuracy':.7781,'precision':.7815,'recall':.7863,'f1':.7839,'roc_auc':.8295}
]).set_index('run')
bert_detail.style.format('{:.2%}')
```

## Embedding training is a separate question
The Word2Vec notebook reports representation-learning accuracy, not sentiment accuracy. It must not be ranked beside classifiers.

```python
word2vec_history = pd.DataFrame({
 'epoch':[1,2,3,4,5],
 'loss':[1.5574,1.3463,1.0070,.6752,.4544],
 'training_accuracy':[.2713,.5312,.7324,.8490,None]})
fig, axes = plt.subplots(1,2,figsize=(10,3.5))
word2vec_history.plot(x='epoch',y='loss',marker='o',ax=axes[0],legend=False,title='Word2Vec loss')
word2vec_history.plot(x='epoch',y='training_accuracy',marker='o',ax=axes[1],legend=False,title='Word2Vec training accuracy')
plt.tight_layout()
```
