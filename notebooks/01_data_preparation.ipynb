# 01 — Tweet data preparation
Reproducible quality checks, normalization and leakage-resistant splitting. The dataset is intentionally not bundled.

```python
import re
import pandas as pd
from sklearn.model_selection import train_test_split

def normalize_tweet(text):
    text = re.sub(r'https?://\S+|www\.\S+', ' URL ', str(text))
    text = re.sub(r'@\w+', ' USER ', text)
    return re.sub(r'\s+', ' ', text).strip()
```
